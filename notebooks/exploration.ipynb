# Experiment: NLP Search Benchmark Exploration

Objective: inspect the sample dataset, run a lightweight lexical baseline, and record the first benchmark metrics.

```python
from __future__ import annotations

from pathlib import Path

from src.dataset import load_documents, load_qrels, load_queries
from src.evaluation import evaluate_run
from src.retrievers.bm25 import BM25Retriever

ROOT = Path.cwd()
DOCUMENTS_PATH = ROOT / "data" / "processed" / "sample_documents.jsonl"
QUERIES_PATH = ROOT / "data" / "processed" / "sample_queries.jsonl"
QRELS_PATH = ROOT / "data" / "qrels" / "sample_qrels.jsonl"
```

## Load Data

The benchmark uses JSONL files for documents, queries, and qrels.

```python
documents = load_documents(DOCUMENTS_PATH)
queries = load_queries(QUERIES_PATH)
qrels = load_qrels(QRELS_PATH)

{
    "documents": len(documents),
    "queries": len(queries),
    "qrels_queries": len(qrels),
}
```

## Run BM25

BM25 is fast enough for quick local checks and does not load transformer models.

```python
retriever = BM25Retriever()
retriever.index(documents)
run = retriever.batch_search(queries, top_k=5)
run
```

## Evaluate

Use the same metrics as the command-line evaluation pipeline.

```python
metrics = evaluate_run(qrels, run, k_values=[5, 10])
metrics
```

- Replace the sample files with your benchmark dataset.
